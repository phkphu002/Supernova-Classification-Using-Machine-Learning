```python
import numpy as np
import matplotlib.pyplot as pl
from sklearn.model_selection import train_test_split
import sncosmo
from astropy.table import Table
```

```python
def fit_supernova(lc):
    bonds = {'z':(0.01, 1.5), 't0':(-100,100),'x0':(-1e-3, 1e-3), 'x1':(-3, 3), 'c':(-0.5, 0.5)}
    model = sncosmo.Model(source='salt2-extended')
    res = sncosmo.fit_lc(lc, model, vparam_names=model.param_names, bounds=bonds, minsnr=0)
    #res, fitted_model = sncosmo.fit_lc(data, model,['z', 't0', 'x0', 'x1', 'c'],bounds={'z':(0.3, 0.7)})
    return res[0].parameters

```

```python
#Importing all files and fitting tto obtain parameters 
k = 0 
params,lyc = [],[]
while(k<=999): #1000 lc
    data = 'test_data/'+str(k)+'.dat' 
    lc = Table.read(data, format='ascii') 
    lyc.append(lc)
    params.append(fit_supernova(lc))
    k += 1
    
params = np.array(params)
#print(params, params.shape)

np.save('testdata.npy',params)
```

Splitting the data set to a training sample and test sample

```python
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import label_binarize
from sklearn.model_selection import train_test_split

#taking X1 and c parameters
X = params[:,3:5] 
labels = np.loadtxt('training_labels.txt')


X_train, X_test, y_train,y_test = train_test_split(X,labels[:,1], test_size=0.5, random_state=42)
```

ML algorithms:
K-Neighbour

```python
from sklearn.feature_selection  import  SelectKBest, chi2
from sklearn import neighbors, datasets
from matplotlib.colors import ListedColormap
from sklearn.neighbors import NearestNeighbors

h = .02  # step size in the mesh
n_neighbors = 20 #X_train)

# Create color maps
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF'])
cmap_bold = ListedColormap(['#FF0000', '#00FF00', '#0000FF'])

X = X_train
y = y_train

nbrs = NearestNeighbors(n_neighbors, algorithm='kd_tree').fit(X)
distances, indices = nbrs.kneighbors(X)

clfk = neighbors.KNeighborsClassifier(n_neighbors, weights='distance')
clfk.fit(X,y)

# Plot the decision boundary.
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))
Z = clfk.predict(np.c_[xx.ravel(), yy.ravel()])

# Put the result into a color plot
Z = Z.reshape(xx.shape)
pl.figure()
pl.pcolormesh(xx, yy, Z, cmap=cmap_light)

# Plot also the training points
pl.scatter(X[:, 0], X[:, 1], c=y, cmap=cmap_bold)
pl.xlim(xx.min(), xx.max())
pl.ylim(yy.min(), yy.max())
pl.title("3-Class classification (k = %i, weights = '%s')"
              % (n_neighbors, 'distance'))

pl.colorbar(ticks=[1,2,3])
pl.show()

pd=clfk.predict(X_test)

print(clfk.predict_proba(X_test)[0:5],'Probability of test')

y_score_k = label_binarize(pd, classes = [0,1,2])

y_test_b = label_binarize(y_test, classes = [0,1,2])
```
